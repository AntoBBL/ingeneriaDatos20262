## 2. Limpieza### 2.1 Valores faltantesEquivalente a `CONTAR.BLANCO` en Excel: contamos cuántos valores faltan por columna.

```python
df.isna().sum().to_frame("valores_faltantes")
```

### 2.2 DuplicadosDetectamos y eliminamos las filas duplicadas (mismo `id_billete`).

```python
duplicados = df.duplicated().sum()
print(f"Filas duplicadas encontradas: {duplicados}")

df = df.drop_duplicates()
print(f"Filas despues de quitar duplicados: {len(df)}")
```

### 2.3 Uniformizar la columna `clase`En el archivo bruto, la clase esta escrita de varias formas (`1era`, `1ra`, `primera`, `2da`, `segunda`, `2nda`). Las agrupamos en dos categorias limpias: `Primera` y `Segunda`.

```python
mapa_clase = {
    "1era": "Primera",
    "1ra": "Primera",
    "primera": "Primera",
    "2da": "Segunda",
    "segunda": "Segunda",
    "2nda": "Segunda",
}

df["clase"] = df["clase"].map(mapa_clase)
df["clase"].value_counts()
```

### 2.4 Precios faltantesMarcamos las filas sin precio para revisarlas aparte, en vez de inventar un valor.

```python
precios_faltantes = df["precio"].isna().sum()
print(f"Precios faltantes: {precios_faltantes}")

df_precio_faltante = df[df["precio"].isna()]
df_precio_faltante[["id_billete", "nombre_cliente", "estado", "precio"]]
```

## 3. TransformaciónConvertimos las fechas de texto a formato fecha real, para poder ordenarlas y filtrarlas.

```python
for col in ["fecha_nacimiento", "fecha_viaje", "fecha_compra"]:
    df[col] = pd.to_datetime(df[col], format="%d/%m/%Y", errors="coerce")

df.dtypes
```

## 4. Analisis — equivalente a las formulas de ExcelCada bloque reproduce en pandas una de las formulas del Excel.

**CONTARA — numero total de billetes registrados**

```python
print(f"Total de billetes: {df['id_billete'].count()}")
```

**CONTAR.SI — billetes por estado**

```python
df["estado"].value_counts()
```

**CONTAR.SI multicriterio — billetes Primera clase Y anulados**

```python
primera_anulados = ((df["clase"] == "Primera") & (df["estado"] == "Anulado")).sum()
print(f"Billetes Primera clase y anulados: {primera_anulados}")
```

**SUMAR.SI — ingreso real (solo billetes pagados)**

```python
ingreso_real = df.loc[df["estado"] == "Pagado", "precio"].sum()
print(f"Ingreso real (COP): {ingreso_real:,.0f}")
```

**Indicador de negocio — billetes pagados por estacion de salida**

```python
df[df["estado"] == "Pagado"].groupby("estacion_salida")["id_billete"].count().sort_values(ascending=False)
```
